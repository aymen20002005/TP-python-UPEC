> Il existe certaines statistiques descriptives simples à calculer qui permettent d'obtenir une meilleure compréhension des données quantitatives. C'est ce qu'on appelle les indicateurs de position. Ils sont au nombre de cinq : moyenne, médiane, quantiles, min, max. La moyenne permet de résumer facilement une liste de données numériques en une seule valeur. Elle se calcule comme suit :

><center>   
$$ \overline{X} = \dfrac{1}{N}\displaystyle\sum\limits_{i=1}^Nx_i $$
</center>

> Où $\overline{X}$ est la moyenne, N la longueur de la liste et $x_i$ les éléments de la liste.

> Deux derniers indicateurs nous permettent d'avoir un bon aperçu : **la variance** et **l'écart-type**. Il font partie de la famille des indicateurs de dispersion : 

* La variance mesure la dispersion autour de la moyenne, c'est à dire à quel point les valeurs présentes dans une liste numérique sont écartées de la moyenne. Elle se définit mathématiquement comme suit :


><center>   
$$ Var(X) = \dfrac{1}{N}\displaystyle\sum\limits_{i=1}^N(x_i-\overline{X})^2 $$
</center>

> Où $\overline{X}$ est la moyenne, N la longueur de la liste et $x_i$ les éléments de la liste.

* L'écart type, quant à lui, se définit mathématiquement comme la racine carrée de la variance. De par sa construction, **l’écart type** est également une mesure de la dispersion des valeurs.
