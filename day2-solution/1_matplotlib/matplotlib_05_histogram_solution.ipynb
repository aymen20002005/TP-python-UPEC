### <i class="fa fa-bar-chart"></i> Histogrammes

> La fonction **``plt.hist``** permet d'afficher des histogrammes. Elle prend principalement en arguments:
- une série de valeurs (``x``) ,
- les bornes des valeurs à utiliser (*range*: par défaut (min(x), max(x)) ) 
- le nombre d'intervalles (*bins*) **ou** les limites explicites des intervalles.<br>

>Dans le cas où le nombre d'intervalles est le seul élément renseigné, ceux-ci seront de tailles égales, mais des intervalles de tailles différentes sont possibles lorsqu'ils sont renseignés explicitement.<br>
Par défaut les intervalles sont ouverts à droite, et fermés à gauche.

> <span style="color:#09b038; text-decoration : underline"> Exemples : </span><br> 
```python
plt.hist([0,1,1, 2, 2, 2, 3, 4, 4, 4, 4, 4, 5, 5], range=(0, 6), bins = 6) 
```

```python
plt.hist([0,8,10,13,15,16,16,18,32,36,39,40,43,45,48,49], bins = [0,10,20,40,50])
```
produit le graphique:

> Pour des histogrammes plus lisibles, le paramètre `rwidth` permet de réduire la largeur des barres avec un espace entre elles (pourcentage entre 0 et 1).<br>


* Créer un histogramme à partir de ``40`` nombres choisis au hasard entre ``0`` et ``10``, grâce à la fonction **`np.random.choice`**, composé de 7 intervalles, d'épaisseur 0.8, et de couleur orange.


````python
# Générer 40 nombres aléatoires entre 0 et 10
x = np.random.choice(np.arange(0, 11), size=40, replace=True)
plt.figure(figsize=(10,6))
plt.hist(x, bins=7, rwidth=0.8, color='orange')
plt.title("Histogramme de valeurs aléatoires")
plt.xlabel("Valeurs")
plt.ylabel("Fréquences")
plt.show()
````

> Il est possible d'afficher les fréquences plutôt que les nombres en ordonnées en ajoutant l'argument `density = True`.<br>
L'ajout de `orientation = 'horizontal'` permet de tracer un histogramme horizontal.<br>

* Afficher un histogramme horizontal à partir de la série de valeurs **``x``** : ``[1, 2, 2, 3, 4, 4, 4, 4, 4, 5, 5]``, avec les nombres entre ``1`` et ``5``, 5 intervalles, des barres réduite de`` 0.6``, de couleur rose (code '#EE3459') et en remplaçant les nombres d'occurences par les fréquences correspondantes.
* Donner à l'axe des abscisses le label 'fréquences', à l'axe des ordonnées le label 'valeurs', et au graphique le titre 'Histogramme horizontal'.


````python
x = [1, 2, 2, 3, 4, 4, 4, 4, 4, 5, 5]
plt.figure(figsize=(10,6))
plt.hist(x, bins=5, range=(1,6), rwidth=0.6, color="#EE3459", orientation='horizontal', density=True)
plt.xlabel('fréquences')
plt.ylabel('valeurs')
plt.title('Histogramme horizontal')
plt.show()
````

> Comme pour les *bar plots* il est possible d'afficher plusieurs histogrammes sur le même graphique, côte à côte ou superposés, en passant comme argument une liste des séries de valeurs. Pour un histogramme avec les séries de valeurs superposées il suffit de rajouter en paramètre ``histtype = 'barstacked'``.

* Lire le fichier `'sales_data.csv'`.

* Créer un histogramme à partir de **`df.Product1`** et **`df.Product2`**, contenant 6 intervalles, et dont les barres sont de couleurs '#f27750' et '#f7bf59'.

* Ajouter les labels correspondants à chaque série. Donner au graphique le titre 'Histogramme 2 series',  à l'axe des abscisses le label 'Ventes' et à l'axe des ordonnées le label 'Fréquences'.

* Afficher les légendes sur le graphique.

````python
# Lecture du fichier CSV et histogrammes pour Product1 et Product2
df = pd.read_csv('sales_data.csv')
plt.figure(figsize=(10,6))
plt.hist([df['Product1'], df['Product2']], bins=6, color=['#f27750', '#f7bf59'], label=['Product1', 'Product2'])
plt.title('Histogramme 2 series')
plt.xlabel('Ventes')
plt.ylabel('Fréquences')
plt.legend()
plt.show()

plt.figure(figsize=(10,6))
plt.hist([df['Product1'], df['Product2']], bins=6, color=['#961939', '#193996'], label=['Product1', 'Product2'], rwidth=0.8, histtype='barstacked')
plt.title('2 series superposées')
plt.xlabel('Ventes')
plt.ylabel('Frequences')
plt.legend()
plt.show()
````

> La méthode `plot.hist()` de Pandas, permet d'afficher un histogramme directement depuis un DataFrame.
> Lorsque plus de deux séries de valeurs sont présentes dans le DataFrame, ou précisées en arguments, les histogrammes sont affichés ensemble au sein du même graphique.
>
>Pour les séparer, il suffit d'ajouter l'argument `subplots=True`, et de régler la disposition des cellules grâce au paramètre `layout`.

> Une seconde méthode, `hist()` permet également d'afficher des histogrammes, mais crée automatiquement un histogramme par colonne, et ne permet pas par exemple différentes couleurs par histogramme.

> Voici quelques exemples d'utilisation :

````python
df.plot.hist(y=['Product1', 'Product2'], bins = 7, rwidth = 0.8 , color= ['#0c4c83', '#830c4c'], alpha=0.5);
df.plot.hist(y=['Product1', 'Product2'], bins = 7, subplots=True, rwidth = 0.8 , color= ['#0c4c83', '#830c4c'], alpha=0.5);

df.hist(column=['Product1', 'Product2'], bins = 7, rwidth = 0.8 , color= ['#0c4c83']);
````
