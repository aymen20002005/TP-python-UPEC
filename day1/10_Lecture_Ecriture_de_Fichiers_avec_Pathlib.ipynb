# Lecture/Écriture de Fichiers avec Pathlib

Cette partie du cours couvre les bases de la manipulation de fichiers et de répertoires en Python en utilisant le module `pathlib`. `Pathlib` offre une approche orientée objet pour gérer les chemins de fichiers, rendant le code plus intuitif et portable. Nous nous concentrerons sur la lecture et l'écriture de fichiers, ainsi que sur le scan de répertoires avec les méthodes `glob` et `rglob`. Nous verrons la syntaxe, des exemples d'utilisation, et quelques exercices pour pratiquer.

En Python, importez `pathlib` avec `from pathlib import Path`. Les objets `Path` représentent des chemins et fournissent des méthodes pour interagir avec le système de fichiers. Contrairement aux strings brutes, `Path` gère les séparateurs (`/` ou `\`) automatiquement selon l'OS.


## 1. Bases de Pathlib : Création et Manipulation de Chemins
Créez un objet `Path` pour représenter un fichier ou un répertoire.

### Exemples
```python
from pathlib import Path

# Chemin courant
chemin_courant = Path.cwd()  # Répertoire de travail actuel
print(chemin_courant)  # Affiche quelque chose comme /home/user/projet

# Création d'un chemin
fichier = Path("exemple.txt")
repertoire = Path("dossier/sous_dossier")

# Joindre des chemins
chemin_complet = repertoire / "fichier.txt"  # Utilise l'opérateur /
print(chemin_complet)  # Affiche dossier/sous_dossier/fichier.txt

# Propriétés
print(fichier.name)  # "exemple.txt"
print(fichier.stem)  # "exemple" (sans extension)
print(fichier.suffix)  # ".txt"
print(fichier.exists())  # True si existe, False sinon
```

Note : `Path` est immutable ; les opérations retournent de nouveaux objets. Utilisez `resolve()` pour un chemin absolu : `fichier.resolve()`.



## 2. Lecture de Fichiers
Utilisez `read_text()` pour les fichiers texte (UTF-8 par défaut) ou `read_bytes()` pour les binaires. Ouvrez avec `open()` pour plus de contrôle.

### Exemples
```python
from pathlib import Path

# Lecture d'un fichier texte
fichier = Path("exemple.txt")
if fichier.exists():
    contenu = fichier.read_text(encoding="utf-8")
    print(contenu)  # Affiche le contenu du fichier

# Lecture binaire (ex. : image)
image = Path("image.png")
if image.exists():
    bytes_data = image.read_bytes()
    print(len(bytes_data))  # Affiche la taille en octets

# Avec open() pour gestion de contexte
with fichier.open("r", encoding="utf-8") as f:
    lignes = f.readlines()
    print(lignes)  # Liste des lignes
```

Note : `open()` est recommandé pour les gros fichiers (lecture ligne par ligne avec `for line in f:`). Gérez les exceptions comme `FileNotFoundError`.



## 3. Écriture de Fichiers
Utilisez `write_text()` pour du texte ou `write_bytes()` pour du binaire. Créez des répertoires parents si nécessaire avec `parent.mkdir(parents=True, exist_ok=True)`.

### Exemples
```python
from pathlib import Path

# Écriture de texte
fichier = Path("nouveau.txt")
fichier.write_text("Bonjour, monde !", encoding="utf-8")
print(fichier.read_text())  # Affiche "Bonjour, monde !"

# Écriture binaire
donnees = b"\x00\x01\x02"
binaire = Path("data.bin")
binaire.write_bytes(donnees)

# Avec open() pour append ou plus
with fichier.open("a", encoding="utf-8") as f:  # Mode append
    f.write("\nLigne supplémentaire.")

# Création de répertoire avant écriture
nouveau_rep = Path("dossier/nouveau_fichier.txt")
nouveau_rep.parent.mkdir(parents=True, exist_ok=True)
nouveau_rep.write_text("Contenu")
```

Note : Les modes pour `open()` : "r" (lecture), "w" (écriture, écrase), "a" (append), "b" (binaire). Utilisez `with` pour fermer automatiquement le fichier.



## 4. Scan de Répertoires avec Glob et Rglob
`glob()` cherche des fichiers matching un pattern dans le répertoire courant. `rglob()` fait de même récursivement (sous-répertoires inclus).

### Exemples
```python
from pathlib import Path

# Glob basique (non récursif)
repertoire = Path(".")
fichiers_txt = list(repertoire.glob("*.txt"))  # Tous les .txt dans le répertoire
for fichier in fichiers_txt:
    print(fichier)  # Affiche les chemins des fichiers .txt

# Rglob récursif
tous_py = list(repertoire.rglob("*.py"))  # Tous les .py dans l'arborescence
print(len(tous_py))  # Nombre de fichiers Python trouvés

# Patterns avancés
images = list(repertoire.glob("image_*.png"))  # image_quelquechose.png
for img in images:
    print(img.name)

# Avec iterdir() pour lister tout (sans pattern)
for element in repertoire.iterdir():
    if element.is_file():
        print(f"Fichier : {element}")
    elif element.is_dir():
        print(f"Répertoire : {element}")
```

Note : `glob` et `rglob` retournent des générateurs ; convertissez en `list()` pour une liste. Patterns : `*` (tout), `?` (un caractère), `[abc]` (un de a,b,c). Utilisez `is_file()` ou `is_dir()` pour filtrer.



## Vérification et Gestion des Erreurs
Vérifiez l'existence avec `exists()`, et gérez les erreurs avec `try-except` :
```python
try:
    contenu = Path("inexistant.txt").read_text()
except FileNotFoundError:
    print("Fichier non trouvé.")
except PermissionError:
    print("Permissions insuffisantes.")
```
