# Les Types de Données en Python

Cette partie du cours couvre les bases des types de données composés en Python, en se concentrant sur les plus courants : `list` (listes), `tuple` (tuples), `dict` (dictionnaires) et `set` (ensembles). Ces types permettent de stocker et de manipuler des collections de données. Contrairement aux types primitifs, ils sont mutables ou immutables selon le cas, et Python gère leur typage dynamiquement.

En Python, vous pouvez créer ces structures directement sans déclaration préalable. Nous verrons des exemples de création, d'opérations courantes et d'utilisation.



## 1. Type `list` (Listes)
Les listes sont des collections ordonnées et mutables de valeurs (qui peuvent être de types différents). Elles sont définies avec des crochets `[]` et supportent les doublons.

### Exemples
```python
# Déclaration d'une liste
fruits = ["pomme", "banane", "orange"]
nombres = [1, 2, 3, 4, 5]
mixte = [42, "texte", True, 3.14]

# Accès et modification
print(fruits[0])  # Affiche "pomme" (indice commence à 0)
fruits[1] = "fraise"  # Modifie l'élément à l'indice 1
print(fruits)  # Affiche ['pomme', 'fraise', 'orange']

# Opérations basiques
fruits.append("kiwi")  # Ajoute à la fin : ['pomme', 'fraise', 'orange', 'kiwi']
fruits.remove("orange")  # Supprime la première occurrence
longueur = len(nombres)  # 5
somme = sum(nombres)  # 15 (pour les nombres)

# Slicing (tranchage)
sous_liste = nombres[1:3]  # [2, 3]
```

Note : Les listes sont mutables, donc modifiables après création.



## 2. Type `tuple` (Tuples)
Les tuples sont des collections ordonnées et immutables (non modifiables après création). Ils sont définis avec des parenthèses `()` et supportent les doublons. Utiles pour des données fixes.

### Exemples
```python
# Déclaration d'un tuple
couleurs = ("rouge", "vert", "bleu")
coordonnees = (10, 20, 30)
singleton = (42,)  # Tuple avec un seul élément (virgule obligatoire)

# Accès
print(couleurs[1])  # Affiche "vert"

# Opérations basiques
longueur = len(coordonnees)  # 3
concat = couleurs + ("jaune",)  # ('rouge', 'vert', 'bleu', 'jaune')
compte = couleurs.count("rouge")  # 1

# Slicing
sous_tuple = coordonnees[0:2]  # (10, 20)
```

Note : Les tuples sont immutables ; tenter de modifier un élément soulève une erreur. Ils sont plus efficaces en mémoire que les listes pour des données constantes.



## 3. Type `dict` (Dictionnaires)
Les dictionnaires sont des collections non ordonnées (avant Python 3.7 ; ordonnées depuis) de paires clé-valeur. Les clés doivent être uniques et immutables (comme des strings ou tuples). Définis avec des accolades `{}`.

### Exemples
```python
# Déclaration d'un dictionnaire
personne = {"nom": "Dupont", "age": 30, "ville": "Paris"}
notes = {"math": 15, "français": 12, "histoire": 18}

# Accès et modification
print(personne["nom"])  # Affiche "Dupont"
personne["age"] = 31  # Modifie la valeur
personne["profession"] = "ingénieur"  # Ajoute une nouvelle paire

# Opérations basiques
cles = list(personne.keys())  # ['nom', 'age', 'ville', 'profession']
valeurs = list(personne.values())  # ['Dupont', 31, 'Paris', 'ingénieur']
del notes["histoire"]  # Supprime la clé
existe = "math" in notes  # True

# Itération
for cle, valeur in personne.items():
    print(f"{cle}: {valeur}")
```

Note : Les clés doivent être hashables (pas de listes comme clés). Utilisez `get()` pour accéder sans erreur si la clé n'existe pas : `personne.get("inconnu", "N/A")`.



## 4. Type `set` (Ensembles)
Les ensembles sont des collections non ordonnées et mutables de valeurs uniques (pas de doublons). Définis avec des accolades `{}` (sans paires) ou `set()`. Utiles pour les opérations mathématiques comme les unions ou intersections.

### Exemples
```python
# Déclaration d'un set
animaux = {"chat", "chien", "oiseau"}
nombres_uniques = set([1, 2, 2, 3])  # {1, 2, 3} (doublons supprimés)

# Opérations basiques
animaux.add("poisson")  # Ajoute un élément
animaux.remove("chien")  # Supprime (erreur si absent)
animaux.discard("inconnu")  # Supprime sans erreur si absent
longueur = len(animaux)  # 3

# Opérations ensemblistes
ensemble1 = {1, 2, 3}
ensemble2 = {3, 4, 5}
union = ensemble1 | ensemble2  # {1, 2, 3, 4, 5}
intersection = ensemble1 & ensemble2  # {3}
difference = ensemble1 - ensemble2  # {1, 2}
```

Note : Les sets ne supportent pas l'indexation (pas d'ordre), et les éléments doivent être hashables.



## Vérification des Types
Pour vérifier le type d'une structure, utilisez la fonction `type()` :
```python
print(type([]))  # <class 'list'>
print(type(()))  # <class 'tuple'>
print(type({}))  # <class 'dict'> (dictionnaire vide)
print(type(set()))  # <class 'set'>
```
