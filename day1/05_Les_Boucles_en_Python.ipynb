# Les Boucles en Python

Cette partie du cours couvre les bases des boucles en Python, en se concentrant sur les instructions `for` et `while`. Les boucles permettent de répéter des blocs de code plusieurs fois, ce qui est essentiel pour l'automatisation et le traitement de collections. Nous verrons la syntaxe, des exemples d'utilisation, et quelques exercices pour pratiquer.

En Python, les boucles utilisent l'indentation pour définir les blocs de code (comme pour les conditions). Vous pouvez aussi utiliser `break` pour sortir prématurément d'une boucle, `continue` pour passer à l'itération suivante, et `else` pour exécuter du code si la boucle se termine normalement (sans `break`).



## 1. Boucle `for`
La boucle `for` itère sur une séquence (comme une liste, un tuple, une string, ou une plage de nombres). Elle est idéale pour un nombre d'itérations connu à l'avance.

### Exemples
```python
# Itération sur une liste
fruits = ["pomme", "banane", "orange"]
for fruit in fruits:
    print(fruit)  # Affiche chaque fruit sur une ligne

# Avec range() pour des nombres
for i in range(5):  # De 0 à 4
    print(i)  # Affiche 0 1 2 3 4

# Range avec paramètres : range(début, fin, pas)
for j in range(2, 10, 2):
    print(j)  # Affiche 2 4 6 8

# Itération sur une string
mot = "Python"
for lettre in mot:
    print(lettre)  # Affiche P y t h o n
```

Note : `range()` génère une séquence de nombres sans créer une liste en mémoire (efficace pour les grandes plages). Vous pouvez itérer sur des dictionnaires avec `.keys()`, `.values()` ou `.items()`.



## 2. Boucle `while`
La boucle `while` répète un bloc tant qu'une condition est vraie. Elle est utile quand le nombre d'itérations est inconnu (dépend d'une condition).

### Exemples
```python
# Compteur simple
compteur = 0
while compteur < 5:
    print(compteur)  # Affiche 0 1 2 3 4
    compteur += 1  # Incrémente

# Avec condition dynamique
solde = 100
while solde > 0:
    print(f"Solde restant : {solde}")
    solde -= 20  # Soustrait 20 à chaque tour

# Boucle infinie (à éviter sans break)
# while True:
#     print("Boucle infinie")  # À stopper manuellement
```

Note : Attention aux boucles infinies si la condition ne devient jamais fausse. Utilisez `break` pour sortir si nécessaire.



## 3. Contrôles de Boucle : `break`, `continue` et `else`
- `break` : Sort immédiatement de la boucle.
- `continue` : Passe à l'itération suivante sans exécuter le reste du bloc.
- `else` : S'exécute si la boucle se termine sans `break` (utile pour des recherches).

### Exemples
```python
# Break et continue
for num in range(10):
    if num == 5:
        break  # Arrête à 5
    if num % 2 == 0:
        continue  # Saute les pairs
    print(num)  # Affiche 1 3 (avant break)

# While avec else
recherche = 7
i = 0
while i < 10:
    if i == recherche:
        print("Trouvé !")
        break
    i += 1
else:
    print("Non trouvé.")  # S'exécute si pas de break
```

Note : Ces contrôles fonctionnent dans les deux types de boucles et peuvent être imbriqués.



## 4. Boucles Imbriquées
Vous pouvez nicher des boucles pour traiter des structures multidimensionnelles (ex. : matrices).

### Exemples
```python
# Table de multiplication
for i in range(1, 4):  # Lignes
    for j in range(1, 4):  # Colonnes
        print(f"{i} * {j} = {i * j}", end=" ")  # Affiche en ligne
    print()  # Saut de ligne
# Affiche :
# 1 * 1 = 1 1 * 2 = 2 1 * 3 = 3
# 2 * 1 = 2 2 * 2 = 4 2 * 3 = 6
# 3 * 1 = 3 3 * 2 = 6 3 * 3 = 9
```

Note : L'indentation doit être précise pour les imbriquements.
