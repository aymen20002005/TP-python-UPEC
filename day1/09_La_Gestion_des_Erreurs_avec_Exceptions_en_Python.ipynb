# La Gestion des Erreurs avec Exceptions en Python

Cette partie du cours couvre les bases de la gestion des erreurs en Python en utilisant les exceptions. Les exceptions sont des événements qui interrompent le flux normal d'exécution d'un programme (ex. : division par zéro, fichier introuvable). Au lieu de planter le programme, vous pouvez les capturer et les gérer gracieusement. Nous verrons la syntaxe des blocs try-except, comment lever des exceptions, créer des exceptions personnalisées, et des bonnes pratiques, avec des exemples et exercices.

En Python, les exceptions sont des objets héritant de `BaseException`. Elles se propagent jusqu'à être capturées ou causer un crash. Utilisez `try-except` pour les gérer.



## 1. Bloc Basique `try-except`
Le bloc `try` contient le code potentiellement erroné ; `except` capture l'exception et exécute un bloc alternatif.

### Exemples
```python
# Try-except simple
try:
    resultat = 10 / 0  # Division par zéro
except ZeroDivisionError:
    print("Erreur : Division par zéro !")

# Avec variable d'exception
try:
    liste = [1, 2, 3]
    print(liste[5])  # Index hors limite
except IndexError as e:
    print(f"Erreur d'index : {e}")
```

Note : Spécifiez le type d'exception (ex. : `ZeroDivisionError`, `IndexError`) pour cibler précisément. Sans type, `except:` capture tout (à éviter car masque les erreurs).



## 2. `except` Multiple, `else` et `finally`
- Plusieurs `except` pour gérer différents types.
- `else` : Exécuté si pas d'exception.
- `finally` : Toujours exécuté (nettoyage, ex. : fermer un fichier).

### Exemples
```python
# Multiples except
try:
    nombre = int("abc")  # ValueError
except ValueError:
    print("Erreur : Conversion impossible.")
except TypeError:
    print("Erreur de type.")

# Avec else et finally
fichier = None
try:
    fichier = open("fichier.txt", "r")
    contenu = fichier.read()
except FileNotFoundError:
    print("Fichier non trouvé.")
else:
    print("Contenu lu avec succès.")
finally:
    if fichier:
        fichier.close()  # Toujours fermer
    print("Fin du bloc try.")
```

Note : `else` est utile pour du code qui ne doit s'exécuter que si tout va bien. `finally` est idéal pour les ressources (fichiers, connexions).

## 3. Lever des Exceptions (`raise`)
Utilisez `raise` pour déclencher manuellement une exception, ou pour re-lancer une capturée.

### Exemples
```python
# Lever une exception
def verifier_age(age):
    if age < 0:
        raise ValueError("L'âge ne peut être négatif.")
    print("Âge valide.")

try:
    verifier_age(-5)
except ValueError as e:
    print(f"Erreur : {e}")

# Re-lancer
try:
    1 / 0
except ZeroDivisionError as e:
    print("Capturée, mais re-lancée.")
    raise  # Re-lance l'exception originale
```

Note : `raise` sans argument re-lance l'exception courante dans un `except`. Ajoutez un message pour plus de clarté.



## 4. Exceptions Personnalisées
Créez vos propres exceptions en héritant de `Exception` pour des erreurs spécifiques à votre application.

### Exemples
```python
# Définition d'une exception personnalisée
class ErreurStockInsuffisant(Exception):
    def __init__(self, message="Stock insuffisant."):
        super().__init__(message)

# Utilisation
def acheter(produit, quantite):
    stock = 10
    if quantite > stock:
        raise ErreurStockInsuffisant(f"Pas assez de {produit} en stock.")
    print("Achat réussi.")

try:
    acheter("pommes", 15)
except ErreurStockInsuffisant as e:
    print(e)
```

Note : Héritez de `Exception` ou d'une sous-classe appropriée. Ajoutez des attributs personnalisés si needed (ex. : quantité demandée).



### Exemple avec Logging
```python
import logging

logging.basicConfig(level=logging.ERROR)
try:
    1 / 0
except ZeroDivisionError:
    logging.error("Division par zéro détectée.")
```

Note : Évitez les `except:` vides ; gérez toujours l'erreur au minimum.
