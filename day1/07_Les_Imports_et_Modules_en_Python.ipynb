# Les Imports et Modules en Python

Cette partie du cours couvre les bases des imports et des modules en Python. Les modules sont des fichiers Python contenant des définitions de variables, fonctions et classes que vous pouvez réutiliser dans d'autres scripts. Les imports permettent d'accéder à ces éléments ou à des bibliothèques externes. Nous verrons comment importer des modules standards, créer vos propres modules, gérer les packages, et des bonnes pratiques, avec des exemples et exercices.

En Python, les modules favorisent la modularité et la réutilisation du code. La bibliothèque standard de Python inclut de nombreux modules utiles (ex. : `math`, `random`). Pour importer, utilisez le mot-clé `import`.



## 1. Import Basique de Modules
L'instruction `import` charge un module entier. Vous accédez à ses éléments avec `module.nom_element`.

### Exemples
```python
# Import d'un module standard
import math

# Utilisation de fonctions du module
print(math.pi)  # Affiche 3.141592653589793
print(math.sqrt(16))  # Affiche 4.0

# Import avec alias (as) pour raccourcir le nom
import random as rd

nombre_aleatoire = rd.randint(1, 10)  # Nombre aléatoire entre 1 et 10
print(nombre_aleatoire)
```

Note : L'alias (`as`) est utile pour les modules aux noms longs ou pour éviter les conflits. Les modules sont chargés une seule fois par session.



## 2. Import Sélectif (from ... import ...)
Importez seulement des éléments spécifiques d'un module avec `from module import element`. Cela évite de préfixer avec le nom du module.

### Exemples
```python
# Import sélectif
from math import sin, cos, pi

# Utilisation directe
angle = pi / 2
print(sin(angle))  # Affiche 1.0
print(cos(angle))  # Affiche 6.123233995736766e-17 (≈0)

# Import de tout (wildcard) - À utiliser avec précaution
from random import *

print(random())  # Nombre flottant aléatoire entre 0 et 1
print(choice(['pomme', 'banane', 'orange']))  # Choisit aléatoirement un fruit
```

Note : L'import wildcard (`*`) peut polluer l'espace de noms et causer des conflits ; préférez les imports sélectifs pour la clarté.



## 3. Création et Import de Vos Propres Modules
Un module est simplement un fichier `.py`. Vous pouvez l'importer depuis le même répertoire ou via `sys.path`.

### Exemples
Supposons un fichier `mon_module.py` :
```python
# Contenu de mon_module.py
def saluer(nom):
    return f"Bonjour, {nom} !"

PI = 3.14159
```

Maintenant, dans un autre fichier `main.py` :
```python
# Import du module personnalisé
import mon_module

print(mon_module.saluer("Alice"))  # Affiche "Bonjour, Alice !"
print(mon_module.PI)  # Affiche 3.14159

# Import sélectif
from mon_module import saluer

print(saluer("Bob"))  # Affiche "Bonjour, Bob !"
```

Note : Pour importer depuis un sous-répertoire, utilisez des points (ex. : `from dossier.sous_module import fonction`). Assurez-vous que le module est dans le chemin de recherche (`sys.path`).



## 4. Packages et Organisation
Un package est un répertoire contenant un fichier `__init__.py` (peut être vide) et des modules. Importez comme un module, avec des points pour la hiérarchie.

### Exemples
Structure de dossier :
```
mon_package/
    __init__.py
    utilitaires.py  # Contient def addition(a, b): return a + b
    constantes.py   # Contient GRAVITE = 9.81
```

Dans un script :
```python
# Import d'un module dans un package
from mon_package.utilitaires import addition

print(addition(5, 3))  # Affiche 8

# Import du package entier
import mon_package.constantes

print(mon_package.constantes.GRAVITE)  # Affiche 9.81
```

Note : `__init__.py` rend le répertoire importable. Vous pouvez y définir des imports par défaut pour simplifier.



## 5. Variable `__name__` et Exécution Conditionnelle
La variable `__name__` vaut `"__main__"` si le script est exécuté directement, sinon le nom du module. Utilisez-la pour du code de test.

### Exemples
Dans `mon_module.py` :
```python
def saluer(nom):
    return f"Bonjour, {nom} !"

if __name__ == "__main__":
    # Code exécuté seulement si lancé directement
    print(saluer("Test"))
```

Si importé ailleurs, le bloc `if` n'exécute pas. Si lancé seul : `python mon_module.py`, il affiche "Bonjour, Test !".

Note : Cela permet de tester un module sans affecter les imports.



## Vérification des Modules
Utilisez `dir()` pour lister les éléments d'un module, ou `__name__` pour vérifier le contexte.

```python
import math
print(dir(math))  # Liste les fonctions et variables de math
print(math.__name__)  # Affiche 'math'
```
