# Les Variables en Python

Cette partie du cours couvre les bases des variables en Python, en se concentrant sur les types primitifs les plus courants : `int` (entiers), `float` (nombres à virgule flottante), `str` (chaînes de caractères) et `bool` (booléens). Nous verrons comment déclarer des variables, des exemples d'utilisation, et quelques exercices pour pratiquer.

En Python, les variables sont des conteneurs pour stocker des données. Vous n'avez pas besoin de déclarer le type explicitement ; Python l'infère automatiquement (typage dynamique). Pour assigner une valeur, utilisez l'opérateur `=`.



## 1. Type `int` (Entiers)
Les entiers représentent des nombres entiers positifs, négatifs ou zéro, sans partie décimale.

### Exemples
```python
# Déclaration d'une variable entière
age = 25  # Positif
temperature = -10  # Négatif
score = 0  # Zéro

# Opérations basiques
somme = age + temperature  # Résultat : 15
produit = age * 2  # Résultat : 50

print(somme)  # Affiche 15
print(produit)  # Affiche 50
```



## 2. Type `float` (Nombres à Virgule Flottante)
Les floats représentent des nombres réels avec une partie décimale. Utilisez un point (.) comme séparateur décimal.

### Exemples
```python
# Déclaration d'une variable float
pi = 3.14159
hauteur = 1.75  # En mètres
solde = -100.50  # Négatif

# Opérations basiques
aire = pi * (2 ** 2)  # Aire d'un cercle de rayon 2 : ≈12.56636
division = hauteur / 2  # Résultat : 0.875

print(aire)  # Affiche ≈12.56636
print(division)  # Affiche 0.875
```

Note : Les floats peuvent avoir des imprécisions dues à la représentation binaire (par exemple, 0.1 + 0.2 = 0.30000000000000004).



## 3. Type `str` (Chaînes de Caractères)
Les strings sont des séquences de caractères, entourées de guillemets simples (') ou doubles (").

### Exemples
```python
# Déclaration d'une variable string
nom = "Alice"
message = 'Bonjour, monde !'
adresse = "123 Rue Exemple"

# Opérations basiques
salutation = "Bonjour, " + nom  # Concaténation : "Bonjour, Alice"
longueur = len(message)  # Longueur : 15

print(salutation)  # Affiche "Bonjour, Alice"
print(longueur)  # Affiche 15
```

Vous pouvez aussi utiliser des triples guillemets pour des strings multilignes :
```python
texte_long = """Ceci est un
texte sur plusieurs lignes."""
```



## 4. Type `bool` (Booléens)
Les booléens représentent des valeurs de vérité : `True` (vrai) ou `False` (faux). Ils sont souvent utilisés dans les conditions.

### Exemples
```python
# Déclaration d'une variable booléenne
est_majeur = True
est_ouvert = False

# Opérations basiques (logiques)
et_logique = est_majeur and est_ouvert  # Résultat : False
ou_logique = est_majeur or est_ouvert  # Résultat : True
negation = not est_majeur  # Résultat : False

print(et_logique)  # Affiche False
print(ou_logique)  # Affiche True
```

Les booléens résultent souvent de comparaisons :
```python
comparaison = (5 > 3)  # True
egalite = (2 == 2)  # True
```



## Vérification des Types
Pour vérifier le type d'une variable, utilisez la fonction `type()` :
```python
print(type(42))  # <class 'int'>
print(type(3.14))  # <class 'float'>
print(type("texte"))  # <class 'str'>
print(type(True))  # <class 'bool'>
```
