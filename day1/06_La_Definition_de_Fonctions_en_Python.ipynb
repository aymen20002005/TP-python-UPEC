# La Définition de Fonctions en Python

Cette partie du cours couvre les bases de la définition et de l'utilisation de fonctions en Python. Les fonctions permettent de regrouper du code réutilisable, d'améliorer la lisibilité et de structurer vos programmes. Nous verrons la syntaxe de base, les paramètres, les retours de valeurs, et des concepts avancés comme les arguments par défaut ou les fonctions anonymes, avec des exemples et exercices.

En Python, une fonction est définie avec le mot-clé `def`, suivi du nom de la fonction, des parenthèses (pour les paramètres), et deux-points `:`. Le bloc de code est indenté. Les fonctions peuvent être appelées (invoquées) n'importe où après leur définition.


## 1. Définition Basique d'une Fonction
Une fonction simple sans paramètres ni retour.

### Exemples
```python
# Définition d'une fonction
def saluer():
    print("Bonjour, monde !")

# Appel de la fonction
saluer()  # Affiche "Bonjour, monde !"

# Fonction avec calcul
def afficher_date():
    annee = 2023
    print(f"Nous sommes en {annee}.")

afficher_date()  # Affiche "Nous sommes en 2023."
```

Note : Les fonctions ne s'exécutent pas tant qu'elles ne sont pas appelées. Le nom doit être descriptif et en snake_case (minuscules avec underscores).



## 2. Paramètres et Arguments
Les paramètres sont des variables dans la définition ; les arguments sont les valeurs passées lors de l'appel. Les fonctions peuvent avoir plusieurs paramètres.

### Exemples
```python
# Fonction avec paramètres
def additionner(a, b):
    resultat = a + b
    print(f"La somme est : {resultat}")

additionner(5, 3)  # Affiche "La somme est : 8"

# Paramètres avec types mixtes
def presenter(nom, age):
    print(f"{nom} a {age} ans.")

presenter("Alice", 25)  # Affiche "Alice a 25 ans."
```

Note : Python n'exige pas de spécifier les types des paramètres (typage dynamique). Les arguments peuvent être passés par position ou par nom (ex. : `presenter(age=25, nom="Alice")`).



## 3. Valeurs de Retour
Utilisez `return` pour renvoyer une valeur de la fonction. Sans `return`, la fonction retourne `None` par défaut.

### Exemples
```python
# Fonction avec retour
def multiplier(x, y):
    return x * y

produit = multiplier(4, 5)  # 20
print(produit)  # Affiche 20

# Retour multiple (sous forme de tuple)
def diviser(a, b):
    quotient = a // b
    reste = a % b
    return quotient, reste

q, r = diviser(10, 3)  # q=3, r=1
print(f"Quotient: {q}, Reste: {r}")
```

Note : `return` termine l'exécution de la fonction. Vous pouvez avoir plusieurs `return` conditionnels.



## 4. Arguments par Défaut et Avancés
- Arguments par défaut : Valeurs prédéfinies si non fournies.
- `*args` : Arguments positionnels variables (tuple).
- `**kwargs` : Arguments nommés variables (dict).

### Exemples
```python
# Arguments par défaut
def saluer_utilisateur(nom="Invité"):
    print(f"Bonjour, {nom} !")

saluer_utilisateur()  # "Bonjour, Invité !"
saluer_utilisateur("Bob")  # "Bonjour, Bob !"

# *args pour un nombre variable d'arguments
def sommer(*args):
    return sum(args)

print(sommer(1, 2, 3))  # 6

# **kwargs pour arguments nommés
def afficher_infos(**kwargs):
    for cle, valeur in kwargs.items():
        print(f"{cle}: {valeur}")

afficher_infos(nom="Jean", age=30)  # Affiche "nom: Jean" et "age: 30"
```

Note : Ordre des paramètres : positionnels, par défaut, *args, **kwargs.



## 5. Fonctions Lambda (Anonymes)
Les lambdas sont des fonctions courtes et anonymes, définies avec `lambda`. Utiles pour des opérations simples, souvent avec `map()`, `filter()`.

### Exemples
```python
# Lambda basique
carre = lambda x: x ** 2
print(carre(4))  # 16

# Avec map()
nombres = [1, 2, 3]
carres = list(map(lambda n: n ** 2, nombres))  # [1, 4, 9]
print(carres)

# Avec filter()
pairs = list(filter(lambda n: n % 2 == 0, [1, 2, 3, 4]))  # [2, 4]
print(pairs)
```

Note : Les lambdas sont limitées à une expression unique, sans statements complexes.



## 6. Portée des Variables (Scope)
- Variables locales : Définies dans la fonction, accessibles seulement dedans.
- Variables globales : Définies en dehors, accessibles partout (utilisez `global` pour modifier dans une fonction).

### Exemples
```python
# Scope local
def test():
    local_var = "Locale"
    print(local_var)

test()  # Affiche "Locale"
# print(local_var)  # Erreur : non définie en dehors

# Scope global
globale = "Globale"
def afficher_global():
    print(globale)  # Accessible

afficher_global()  # Affiche "Globale"
```

Note : Préférez les variables locales pour éviter les effets de bord.



## Vérification des Types
Les fonctions sont des objets de type `function` :
```python
def ma_fonc():
    pass

print(type(ma_fonc))  # <class 'function'>
```
