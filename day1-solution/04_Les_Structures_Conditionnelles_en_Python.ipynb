# Les Structures Conditionnelles en Python

Cette partie du cours couvre les bases des structures conditionnelles en Python, en se concentrant sur les instructions `if`, `else`, `elif` et les expressions ternaires. Ces structures permettent d'exécuter du code en fonction de conditions booléennes, rendant vos programmes plus dynamiques et décisionnels. Nous verrons la syntaxe, des exemples d'utilisation, et quelques exercices pour pratiquer.

En Python, les conditions sont évaluées à `True` ou `False`. Utilisez des opérateurs de comparaison (`==`, `!=`, `>`, `<`, `>=`, `<=`) et logiques (`and`, `or`, `not`). L'indentation (généralement 4 espaces) définit les blocs de code.



## 1. Instruction `if`
L'instruction `if` exécute un bloc de code si une condition est vraie.

### Exemples
```python
# Condition simple
age = 18
if age >= 18:
    print("Vous êtes majeur.")  # S'exécute si vrai

# Avec opérateurs logiques
note = 15
if note >= 10 and note < 15:
    print("Mention passable.")

# Sans exécution si faux
if note < 10:
    print("Redoublement.")  # Ne s'exécute pas ici
```

Note : Si la condition est fausse, le bloc est ignoré. Pas de parenthèses obligatoires autour de la condition, mais possibles pour la clarté.



## 2. Instruction `if-else`
Ajoutez `else` pour exécuter un bloc alternatif si la condition `if` est fausse.

### Exemples
```python
# if-else basique
temperature = 25
if temperature > 20:
    print("Il fait chaud.")
else:
    print("Il fait frais.")

# Avec booléens
est_pluie = False
if est_pluie:
    print("Prenez un parapluie.")
else:
    print("Bonne promenade !")
```

Note : `else` n'a pas de condition ; il s'exécute toujours si `if` est faux.



## 3. Instruction `if-elif-else`
Utilisez `elif` (else if) pour tester plusieurs conditions en chaîne. `else` est optionnel à la fin.

### Exemples
```python
# Chaîne de conditions
score = 85
if score >= 90:
    print("Excellent !")
elif score >= 80:
    print("Très bien.")
elif score >= 70:
    print("Bien.")
else:
    print("À améliorer.")

# Plusieurs elif
jour = "lundi"
if jour == "lundi":
    print("Début de semaine.")
elif jour == "vendredi":
    print("Fin de semaine.")
elif jour in ["samedi", "dimanche"]:
    print("Week-end !")
else:
    print("Milieu de semaine.")
```

Note : Les conditions sont testées séquentiellement ; la première vraie exécute son bloc, et les suivantes sont ignorées.



## 4. Expressions Ternaires (Condition en Ligne)
Une forme compacte de `if-else` pour assigner des valeurs en une ligne.

### Exemples
```python
# Ternaire basique
age = 16
statut = "majeur" if age >= 18 else "mineur"
print(statut)  # Affiche "mineur"

# Avec calcul
x = 10
parite = "pair" if x % 2 == 0 else "impair"
print(parite)  # Affiche "pair"

# Imbriquée (à utiliser avec modération pour la lisibilité)
note = 12
mention = "excellent" if note >= 16 else "bien" if note >= 12 else "passable"
print(mention)  # Affiche "bien"
```

Note : Syntaxe : `valeur_si_vrai if condition else valeur_si_faux`. Idéal pour des assignations simples, mais évitez les imbriquements complexes.



## 5. Bonnes Pratiques et Astuces
- **Indentation** : Toujours cohérente ; mélangez pas espaces et tabs.
- **Conditions composées** : Utilisez `and`/`or` pour combiner, et `not` pour inverser.
- **Vérité et Fausseté** : En Python, `0`, `""`, `[]`, `None` sont faux ; le reste est vrai.
- **if imbriqués** : Possibles, mais préférez `elif` pour la lisibilité.
```python
# Exemple d'imbriqué
if age >= 18:
    if est_etudiant:
        print("Réduction étudiante.")
    else:
        print("Tarif adulte.")
else:
    print("Tarif enfant.")
```



## Exercices
Voici quelques exercices simples pour pratiquer. Essayez de les résoudre dans un interpréteur Python, puis vérifiez les résultats.

1. **Exercice 1 : Condition simple**
   Déclarez `heure = 14`. Utilisez `if` pour afficher "Bonjour" si `heure < 18`, sinon rien.



````python
heure = 14

print("Bonjour") if heure < 18 else None

if heure < 18:
    print("Bonjour")
````

2. **Exercice 2 : if-else**
   Déclarez `nombre = 7`. Affichez "Positif" si `nombre > 0`, "Négatif" si `nombre < 0`, sinon "Zéro" (utilisez if-elif-else).



````python
nombre = 0

if nombre > 0:
    print("Positif")
else:
    if nombre < 0:
        print("Négatif")
    else:
        print("Zéro")


print("Positif") if nombre > 0 else print("Négatif") if nombre < 0 else print("Zéro")

````

3. **Exercice 3 : Chaîne de conditions**
   Déclarez `note = 14`. Utilisez if-elif-else pour afficher "A" (>=16), "B" (>=12), "C" (>=10), sinon "Echec".



````python
note = 16

if note >= 16:
    print("A")

elif note >= 12:
    print("B")

elif note >= 10:
    print("C")

else:
    print("Echec")
````

4. **Exercice 4 : Ternaire**
   Déclarez `pluie = True`. Utilisez une expression ternaire pour assigner à `activite` "Cinéma" si pluie, sinon "Promenade". Affichez `activite`.



````python
pluie = True

activite = "Cinéma" if pluie else "Promenade"

print(activite)
````

5. **Exercice 5 : Imbriqué et logique**
   Déclarez `age = 20`, `permis = False`. Utilisez des if imbriqués ou logiques pour afficher "Peut conduire" si age >=18 et permis=True, sinon des messages appropriés (ex. "Trop jeune" ou "Besoin de permis").


````python
age = 20
permis = False

if age > 18:
    if permis:
        print("Peu conduire")
    else:
        print("Besoin de permis")
else:
    print("Trop jeune pour conduire")

````

````python
def prix_voiture(voiture: str) -> float:
    if voiture == "Renault":
        return 10_000

    elif voiture == "BMW":
        return 15_000

    elif voiture == "PG":
        return 12_500


prix_voiture("PG")
````

````python
BASE_PRIX_VOITURE: dict[str, float] = {
    "Renault": 10_000,
    "BMW": 15_000,
    "PG": 12_500,
    }

def prix_voiture(voiture: str) -> float:
    return BASE_PRIX_VOITURE[voiture]


prix_voiture("PG")
````
