# Les Classes en Python

Cette partie du cours couvre les bases des classes en Python, qui sont au cœur de la programmation orientée objet (POO). Les classes permettent de créer des objets personnalisés avec des attributs (données) et des méthodes (fonctions). Nous verrons la définition d'une classe, les instances, les méthodes, l'héritage, et des concepts comme les attributs de classe vs. d'instance, avec des exemples et exercices.

En Python, une classe est définie avec le mot-clé `class`, suivi du nom de la classe (en CamelCase par convention), et deux-points `:`. Le bloc est indenté. Les classes encapsulent des données et des comportements, favorisant la réutilisation et l'organisation du code.



## 1. Définition Basique d'une Classe
Une classe simple avec des attributs et une méthode.

### Exemples
```python
# Définition d'une classe
class Livre:
    # Attribut de classe (partagé par toutes les instances)
    type_support = "Papier"

    # Méthode d'initialisation (constructeur)
    def __init__(self, titre, pages):
        # Attributs d'instance (propres à chaque objet)
        self.titre = titre
        self.pages = pages

    # Méthode d'instance
    def afficher_info(self):
        print(f"Le livre '{self.titre}' a {self.pages} pages.")

# Création d'instances (objets)
livre1 = Livre("Python pour Débutants", 300)
livre2 = Livre("Apprendre la POO", 250)

# Accès aux attributs et méthodes
print(livre1.titre)  # Affiche "Python pour Débutants"
print(livre1.type_support)  # Affiche "Papier" (attribut de classe)
livre1.afficher_info()  # Affiche "Le livre 'Python pour Débutants' a 300 pages."
```

Note : `__init__` est appelée automatiquement lors de la création d'une instance. `self` fait référence à l'instance actuelle (comme `this` dans d'autres langages).



````python
class Operation:
    def __init__(self):
        pass

    def add(self, a, b):
        return a+b

    def sub(self, a, b):
        return a-b

    def mul(self, a, b):
        return a*b

operations = Operation()
print(operations.sub(10, 5))
print(operations.mul(5, 5))
print(operations.add(2, 6))
````

## 2. Attributs et Méthodes
- **Attributs d'instance** : Définis dans `__init__`, uniques à chaque objet.
- **Attributs de classe** : Définis directement dans la classe, partagés.
- **Méthodes d'instance** : Prennent `self` comme premier paramètre.
- **Méthodes de classe** : Décorées avec `@classmethod`, prennent `cls`.
- **Méthodes statiques** : Décorées avec `@staticmethod`, sans `self` ou `cls`.

### Exemples
```python
class Ordinateur:
    # Attribut de classe
    systeme_exploitation = "Windows"

    def __init__(self, marque, ram):
        self.marque = marque
        self.ram = ram
        self.stockage = 512  # Attribut d'instance

    # Méthode d'instance
    def augmenter_ram(self, ajout):
        self.ram += ajout
        print(f"RAM totale : {self.ram} Go")

    # Méthode de classe
    @classmethod
    def description(cls):
        return f"Un ordinateur utilise {cls.systeme_exploitation} par défaut."

    # Méthode statique
    @staticmethod
    def alerte():
        print("Attention : mise à jour disponible !")

# Utilisation
pc = Ordinateur("Dell", 8)
pc.augmenter_ram(4)  # Affiche "RAM totale : 12 Go"
print(Ordinateur.description())  # Affiche "Un ordinateur utilise Windows par défaut."
Ordinateur.alerte()  # Affiche "Attention : mise à jour disponible !"
```

Note : Les méthodes de classe peuvent créer des instances alternatives ; les statiques sont comme des fonctions utilitaires liées à la classe.



## 3. Héritage
L'héritage permet à une classe enfant d'hériter des attributs et méthodes d'une classe parent. Utilisez `super()` pour appeler les méthodes du parent.

### Exemples
```python
# Classe parent
class Animal:
    def __init__(self, nom):
        self.nom = nom

    def manger(self):
        print(f"{self.nom} mange.")

# Classe enfant (hériite de Animal)
class Chat(Animal):
    def __init__(self, nom, couleur):
        super().__init__(nom)  # Appel au constructeur parent
        self.couleur = couleur

    # Surcharge (override) d'une méthode
    def manger(self):
        print(f"{self.nom} mange des croquettes.")

    # Nouvelle méthode
    def miauler(self):
        print("Miaou !")

# Utilisation
chat = Chat("Minou", "gris")
chat.manger()  # Affiche "Minou mange des croquettes." (surchargé)
chat.miauler()  # Affiche "Miaou !"
```

Note : Python supporte l'héritage multiple (ex. : `class Enfant(Parent1, Parent2)`), mais utilisez-le avec précaution pour éviter les conflits.



## 4. Encapsulation et Propriétés
L'encapsulation cache les détails internes. Utilisez `_` pour "protégé" (convention), `__` pour "privé" (name mangling). Les propriétés gèrent l'accès avec `@property`.

### Exemples
```python
class CompteBancaire:
    def __init__(self, solde_initial):
        self._solde = solde_initial  # Protégé

    @property
    def solde(self):
        return self._solde

    @solde.setter
    def solde(self, valeur):
        if valeur < 0:
            print("Solde ne peut être négatif.")
        else:
            self._solde = valeur

    def deposer(self, montant):
        self.solde += montant

# Utilisation
compte = CompteBancaire(100)
print(compte.solde)  # Affiche 100 (via getter)
compte.solde = -50  # Affiche "Solde ne peut être négatif."
compte.deposer(50)
print(compte.solde)  # Affiche 150
```

Note : Pas de vrai privé en Python, mais les conventions aident à la maintenance.



## 5. Méthodes Spéciales (Dunder Methods)
Les méthodes comme `__str__`, `__repr__`, `__len__` personnalisent le comportement (ex. : pour `print()` ou `len()`).

### Exemples
```python
class Livre:
    def __init__(self, titre, pages):
        self.titre = titre
        self.pages = pages

    def __str__(self):
        return f"Livre : {self.titre}"

    def __len__(self):
        return self.pages

livre = Livre("Python Avancé", 300)
print(livre)  # Affiche "Livre : Python Avancé" (via __str__)
print(len(livre))  # Affiche 300
```

Note : `__repr__` est pour une représentation technique (debug), `__str__` pour l'utilisateur.



## Vérification des Types
Utilisez `type()` ou `isinstance()` :
```python
print(type(chien1))  # <class '__main__.Chien'>
print(isinstance(chien1, Chien))  # True
```

3. **Exercice 3 : Héritage**
   Créez une classe `Employe` héritant de `Personne` (ex. 1), ajoutez un attribut `salaire` et surchargez `se_presenter()` pour inclure le salaire.



````python
class Personne:
    def __init__(self, nom, age):
        self.nom = nom
        self.age = age

    def se_presenter(self):
        print(f"Je m'appelle {self.nom} et j'ai {self.age} ans.")

    def __str__(self):
        return f"Je m'appelle {self.nom} et j'ai {self.age} ans."


class Employe(Personne):
    def __init__(self, nom, age, salaire):
        super().__init__(nom, age)
        self.salaire = salaire

    def se_presenter(self):
        print(f"Je m'appelle {self.nom} et j'ai {self.age} ans, mon salaire est de {self.salaire} euros.")

bob = Employe("Bob", 30, 2000)
bob.se_presenter()
print(bob)
````

4. **Exercice 4 : Propriétés**
   Dans une classe `Cercle`, utilisez une propriété pour calculer l'aire (`pi * rayon ** 2`) en lecture seule. Initialisez avec `rayon`.



````python
import math

class Cercle:
    def __init__(self, rayon):
        self.rayon = rayon
        # self.pi = 3.14

    @property
    def aire(self):
        return math.pi * self.rayon ** 2

cercle = Cercle(rayon=2)
print(cercle.aire)
````
