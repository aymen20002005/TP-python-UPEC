# Les Type Hints et le Typage Statique en Python

Cette partie du cours couvre les bases des type hints (annotations de type) et du typage statique en Python. Les type hints, introduits dans Python 3.5 via PEP 484, permettent d'indiquer les types attendus pour les variables, paramètres et retours de fonctions, améliorant la lisibilité et facilitant la détection d'erreurs sans affecter l'exécution (Python reste dynamique). Le typage statique utilise des outils comme mypy pour vérifier ces hints avant runtime. Nous verrons la syntaxe, des exemples d'utilisation, et quelques exercices pour pratiquer.

En Python, les type hints sont optionnels et n'empêchent pas les erreurs runtime, mais ils aident les IDE (comme VS Code) pour l'autocomplétion et les linters pour les checks statiques. Importez `typing` pour des types complexes.



## 1. Annotations Basiques pour Variables et Fonctions
Ajoutez des annotations avec `:` après le nom, et `->` pour le retour d'une fonction.

### Exemples
```python
# Annotation pour une variable
age: int = 25
nom: str = "Alice"

# Fonction avec annotations
def additionner(a: int, b: int) -> int:
    return a + b

resultat: int = additionner(5, 3)  # Inféré comme int, mais annoté pour clarté

# Avec valeur par défaut
def saluer(prenom: str = "Invité") -> str:
    return f"Bonjour, {prenom} !"

print(saluer())  # Affiche "Bonjour, Invité !"
```

Note : Les annotations n'ont aucun effet runtime ; `age = "vingt-cinq"` exécute sans erreur, mais un outil statique comme mypy le signalera.



## 2. Types Complexes avec `typing`
Utilisez le module `typing` pour des types comme listes, dictionnaires, optionnels, unions, etc.

### Exemples
```python
from typing import List, Dict, Optional, Union, Tuple

# Liste d'entiers
nombres: List[int] = [1, 2, 3]

# Dictionnaire str -> float
notes: Dict[str, float] = {"math": 15.5, "français": 12.0}

# Optional (peut être None)
adresse: Optional[str] = None

# Union (plusieurs types possibles)
identifiant: Union[int, str] = "user123"  # Ou 123

# Tuple avec types spécifiques
coordonnees: Tuple[float, float] = (3.14, 42.0)

# Fonction avec types complexes
def traiter_donnees(data: List[Dict[str, Union[int, str]]]) -> Optional[int]:
    if not data:
        return None
    return len(data)

exemple: List[Dict[str, Union[int, str]]] = [{"id": 1, "nom": "Bob"}]
print(traiter_donnees(exemple))  # Affiche 1
```

Note : Depuis Python 3.10, utilisez des builtins comme `list[int]` au lieu de `List[int]` (plus simple, pas besoin d'importer `typing`). `Any` pour tout type, `NoReturn` pour fonctions qui ne retournent rien (ex. : raise).



## 3. Typage pour Classes et Méthodes
Annotez les attributs dans `__init__` et les méthodes. Utilisez `Self` (depuis 3.11) pour le type de l'instance.

### Exemples
```python
from typing import Self  # Depuis Python 3.11

class Point:
    def __init__(self, x: int, y: int) -> None:
        self.x: int = x
        self.y: int = y

    def distance(self, other: Self) -> float:
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5

p1: Point = Point(0, 0)
p2: Point = Point(3, 4)
print(p1.distance(p2))  # Affiche 5.0
```

Note : Pour les dataclasses (voir cours précédent), les annotations sont utilisées automatiquement pour générer `__init__` etc.



## 4. Typage Statique avec Outils comme Mypy
Installez mypy (`pip install mypy`) et exécutez `mypy fichier.py` pour vérifier les types statiquement. Il détecte les incohérences sans exécuter le code.

### Exemples
Supposons un fichier `exemple.py` :
```python
def multiplier(a: int, b: str) -> int:  # Erreur : b devrait être int pour multiplication
    return a * b

multiplier(5, "trois")  # Mypy signalera l'erreur sur b: str
```

Exécution : `mypy exemple.py` affichera des erreurs comme "Incompatible type for 'b' (expected 'int', got 'str')".

Note : Autres outils : pyright (de Microsoft, intégré à VS Code), pytype (de Google). Configurez-les dans votre IDE pour des checks en temps réel.



## 5. Avantages et Bonnes Pratiques
- **Avantages** : Meilleure documentation, détection précoce d'erreurs, meilleure autocomplétion dans les IDE, facilite la collaboration.
- **Pratiques** : Commencez par annoter les fonctions publiques. Utilisez `TypeAlias` pour des types réutilisables (ex. : `Coord = Tuple[int, int]`). Évitez les annotations excessives pour du code simple.

### Exemple Avancé : TypeAlias et Generics
```python
from typing import TypeAlias, Generic, TypeVar

T = TypeVar("T")
class Boite(Generic[T]):
    def __init__(self, valeur: T) -> None:
        self.valeur: T = valeur

Coord: TypeAlias = Tuple[int, int]
position: Coord = (10, 20)
```

   Définissez une fonction `concatener(a: str, b: str) -> str` qui retourne a + b. Annotez une variable `resultat: str` et appelez-la.





````python
def concatener(a: str, b: str) -> str:
    return a + b

resultat: str = concatener("Alice", " Bob")
print(resultat)
````

2. **Exercice 2 : Types complexes**
   Déclarez `utilisateurs: Dict[str, List[int]] = {"Alice": [25, 30]}`. Écrivez une fonction qui prend ce dict et retourne Optional[List[int]] pour une clé donnée.



````python
from typing import Dict, List, Optional

utilisateurs: Dict[str, List[int]] = {"Alice": [25, 30]}

def foo(utilisateur: Dict[str, List[int]], key:str) -> Optional[List[int]]:
    return utilisateur.get(key, None)
````

````python
utilisateurs: dict[str, list[int]] = {"Alice": [25, 30]}

def foo(utilisateur: dict[str, list[int]], key: str) -> list[int] | None:
    return utilisateur.get(key, None)

print(foo(utilisateurs, "Bob"))
print(foo(utilisateurs, "Alice"))
````

3. **Exercice 3 : Classe avec hints**
   Créez une classe `Employe` avec `__init__(nom: str, salaire: float)` et une méthode `augmenter(salaire: float) -> None` qui ajoute au salaire.



````python
class Employe:
    def __init__(self, nom: str, salaire: float) -> None:
        self.nom: str = nom
        self.salaire: float = salaire

        def augmenter(self, salaire: float) -> None:
            self.salaire += salaire
````
