# Les Opérations et Manipulation de Chaînes en Python

Cette partie du cours rapide couvre les bases des opérations et de la manipulation de chaînes de caractères (`str`) en Python. Les strings sont immutables (non modifiables directement), mais vous pouvez créer de nouvelles strings via des opérations. Nous explorerons la concaténation, le slicing, les méthodes courantes, le formatage, et plus encore, avec des exemples et exercices.

En Python, les strings sont des séquences de caractères, et de nombreuses opérations s'appliquent comme sur les listes (indexation, itération). Utilisez des guillemets simples, doubles ou triples pour les définir.


## 1. Concaténation et Répétition
La concaténation joint des strings avec `+`, et la répétition utilise `*`.

### Exemples
```python
# Concaténation
prenom = "Jean"
nom = "Dupont"
nom_complet = prenom + " " + nom  # "Jean Dupont"
salutation = "Bonjour, " + nom_complet + " !"  # "Bonjour, Jean Dupont !"

# Répétition
separateur = "-" * 10  # "----------"
echo = "hello" * 3  # "hellohellohello"

print(nom_complet)  # Affiche "Jean Dupont"
print(separateur)  # Affiche "----------"
```

Note : Pour une concaténation efficace avec de nombreuses strings, utilisez `join()` (voir section 3).



## 2. Indexation et Slicing
Les strings sont indexées (premier caractère à l'indice 0). Le slicing extrait des sous-chaînes avec `[début:fin:pas]`.

### Exemples
```python
texte = "Python est génial !"

# Indexation
premier = texte[0]  # "P"
dernier = texte[-1]  # "!" (indice négatif pour la fin)

# Slicing
sous_texte = texte[0:6]  # "Python" (de 0 à 5 inclus)
inverse = texte[::-1]  # "! laiénég tse nohtyP" (inversion)
pas_de_deux = texte[::2]  # "Pto sgnil!" (tous les 2 caractères)

print(premier)  # Affiche "P"
print(sous_texte)  # Affiche "Python"
print(inverse)  # Affiche "! laiénég tse nohtyP"
```

Note : Les strings sont immutables ; `texte[0] = "p"` soulève une erreur. Créez une nouvelle string pour modifier.



## 3. Méthodes Courantes de Manipulation
Python offre de nombreuses méthodes intégrées pour les strings. Voici les plus utiles :

- `upper()`, `lower()`, `capitalize()`, `title()` : Changement de casse.
- `strip()`, `lstrip()`, `rstrip()` : Suppression d'espaces (ou caractères spécifiés).
- `split()`, `rsplit()` : Division en liste.
- `join()` : Jointure d'une liste en string.
- `find()`, `index()`, `rfind()` : Recherche de sous-chaîne.
- `replace()` : Remplacement.
- `startswith()`, `endswith()` : Vérification de début/fin.
- `isalpha()`, `isdigit()`, `isalnum()` : Tests de contenu.

### Exemples
```python
phrase = "  Bonjour le Monde !  "

# Casse
maj = phrase.upper()  # "  BONJOUR LE MONDE !  "
minusc = phrase.lower()  # "  bonjour le monde !  "
titre = phrase.title()  # "  Bonjour Le Monde !  "

# Nettoyage
nettoye = phrase.strip()  # "Bonjour le Monde !"

# Division et jointure
mots = nettoye.split()  # ['Bonjour', 'le', 'Monde', '!']
rejoint = "-".join(mots)  # "Bonjour-le-Monde-!"

# Recherche et remplacement
position = nettoye.find("Monde")  # 11 (première occurrence)
remplace = nettoye.replace("Monde", "Python")  # "Bonjour le Python !"

# Tests
est_alpha = "abc".isalpha()  # True
est_chiffre = "123".isdigit()  # True

print(maj)  # Affiche "  BONJOUR LE MONDE !  "
print(rejoint)  # Affiche "Bonjour-le-Monde-!"
print(position)  # Affiche 11
```

Note : La plupart des méthodes retournent une nouvelle string sans modifier l'originale.



## 4. Formatage de Strings
Outre la concaténation, Python propose plusieurs façons de formater : `%` (ancien), `format()`, et f-strings (moderne, recommandé).

### Exemples
```python
nom = "Alice"
age = 25

# Méthode format()
formate = "Bonjour, {} ! Tu as {} ans.".format(nom, age)
# "Bonjour, Alice ! Tu as 25 ans."

# F-strings (depuis Python 3.6)
f_format = f"Bonjour, {nom} ! Tu as {age} ans."
# Identique

# Formatage avancé avec f-strings
pi = 3.14159
precise = f"Pi : {pi:.2f}"  # "Pi : 3.14"
calcule = f"Double de {age} : {age * 2}"  # "Double de 25 : 50"

print(formate)  # Affiche "Bonjour, Alice ! Tu as 25 ans."
print(precise)  # Affiche "Pi : 3.14"
```

Note : Les f-strings sont les plus lisibles et performantes. Référez-vous au cours précédent pour plus sur les f-strings.



## 5. Échappement et Strings Brutes
Utilisez `\` pour échapper des caractères spéciaux (ex. : `\"` pour un guillemet). Les strings brutes (`r"..."`) ignorent les échappements.

### Exemples
```python
echappe = "Ceci est une \"citation\"."  # Avec échappement
brut = r"C:\Users\nouveau"  # Pas d'interprétation de \n comme saut de ligne

print(echappe)  # Affiche Ceci est une "citation".
print(brut)  # Affiche C:\Users\nouveau
```
