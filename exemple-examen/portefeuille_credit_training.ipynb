def get_defaut_rate(self, group_col):
    """Calcule le taux de défaut (Oui/Total) pour chaque groupe"""
    rates = {}
    
    for group_name, group_data in self.df.groupby(group_col):
        total = len(group_data)
        defauts = len(group_data[group_data[self.target] == 'Oui'])
        rate = defauts / total if total > 0 else 0
        rates[group_name] = rate
    
    # Trier par taux décroissant
    rates = dict(sorted(rates.items(), key=lambda x: x[1], reverse=True))
    return rates

CreditAnalyzer.get_defaut_rate = get_defaut_rate
