## Question 6 : Création d'une Fonction pour Imputer les Valeurs Manquantes
### Énoncé
Créez une fonction `impute_missing` qui prend `df` et une colonne. Si la colonne est numérique ('float64' ou 'int64'), imputez avec la médiane ; sinon, avec le mode.

```python
# Votre code pour la Question 6

def impute_missing(df,col):
    if df[col].dtype in ['float64', 'int64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])
    return df
```

## Question 7 : Application de l'Imputation sur l'Âge

```python
# Votre code pour la Question 7
df = impute_missing(df,'Age')
print(df['Age'].isnull().sum())  # Vérifie qu'il n'y a plus de valeurs manquantes dans la colonne 'Age'
```

## Question 10 : Calcul Manuel de la Moyenne d'Âge avec une Boucle
### Énoncé
Créez une fonction `manual_mean_age` qui utilise une boucle `for` sur `df['Age']` pour calculer la moyenne. Affichez le résultat.


```python
# Votre code pour la Question 10
def manual_mean_age(column):
    total = 0
    count = 0
    for value in column:
        if pd.notnull(value):  # Vérifie si la valeur n'est pas manquante
            total += value
            count += 1
    return total / count if count > 0 else None  # Retourne la moyenne si
```

```python
print(manual_mean_age(df['Age']))  # Affiche la moyenne calculée manuellement pour la colonne 'Age'
```
