````python
# Interface Segregation Principle

from abc import ABC, abstractmethod
class parent(ABC):
    @abstractmethod
    def printf(self):
        print("parent")
    def argent(self):
        print("argent")
    def travail(self):
        print("travail")

class fils(parent):
    def travail(self):
            print("travail")

fils = fils()
fils.printf()  # This will call the printf function from the fils class and print "fils"
````

```python
    else:
        return 'Invalid op'
```

Tâche : Refactorer avec un dictionary mapping pour opérations, respectant Open-Closed.

````python
# Insérez votre code ici

dictionnaire_operation = {
    "addition": lambda x, y: x + y,
    "soustraction": lambda x, y: x - y,
    "multiplication": lambda x, y: x * y
}

def calculate(operation, x, y):
    if operation in dictionnaire_operation:
        return dictionnaire_operation[operation](x, y)
    else:
        raise ValueError(f"Unknown operation: {operation}")

print(calculate("addition", 5, 3))  # Output: 8
````

### Exercice 8 : Code dupliqué

Deux fonctions similaires pour traiter des listes. Violation de DRY (Don't Repeat Yourself).

Code mal écrit :

```python
def sum_even(numbers):
    total = 0
    for n in numbers:
        if n % 2 == 0:
            total += n
    return total

def sum_odd(numbers):
    total = 0
    for n in numbers:
        if n % 2 != 0:
            total += n
    return total
```

Tâche : Refactorer en une fonction générique.

````python
# Insérez votre code ici


def sum_all(numbers, condition):
    total = 0
    for n in numbers:
        if condition(n):
            total += n
    return total

sum_odd = sum_all([1, 2, 3, 4, 5], lambda x: x % 2 != 0)
print(sum_odd)
sum_even = sum_all([1, 2, 3, 4, 5], lambda x: x % 2 == 0)
print(sum_even)
````
